## Задача 1


$F(x) = \left\{\begin{matrix}
0, \text{если } x < 0 \\
\frac{x^2}{2}, \text{если } 0 \leq x \leq \sqrt{2}\\
1, \text{если }  x > \sqrt{2}
\end{matrix}\right.$

1. Постройте график данной функции распределения.
2. Найдите вероятность того, что Х примет значение, принадлежащее интервалу $X \in (0.2;1)$.

```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
import scipy as sc
def F(x):
    return np.where(x < 0, 0,
            np.where(x <= np.sqrt(2), (x**2)/2, 1))

x = np.linspace(-2, 2, 100)
plt.plot(x, F(x))
plt.show()

display(f"Вероятность: {F(1) - F(0.2)}")
```

## Задача 2

Постройте данные функции и определите, являются ли они плотностью распределения некоторой случайной величины:
1. $f(x) = \dfrac{x}{\pi(1+x^2)},\ x \in (-\inf,+\inf)$
2. $f(x) = \left\{\begin{matrix}
\dfrac{1}{2},\ x \in (-1;1] \\
0,\ x \notin (-1;1] \end{matrix}\right.$
3. $f(x) = \left\{\begin{matrix}
0, \text{ если } x < 0\  \&\  x > 2 \\
ax^2, \text{ если } 0 \leq x \leq 2
\end{matrix}\right.$

```python
x = np.linspace(-100, 100, 300)
f1 = x/(np.pi*(1+x**2))


plt.plot(x, f1)
plt.show()

#бесконечность построить нельзя, но на таком интервале уже видно, что не выполняется главный признак функции распределения - ее значения меньше 0
```

```python
x = np.linspace(-1.5, 1.5, 10)
f1 = np.where((x > -1) & (x <= 1), 1/2, 0)
xsp = sp.symbols('x')
f1sp = sp.Piecewise((1/2, sp.And(-1 < xsp, xsp <= 1)))
plt.plot(x, f1)
plt.show()
sp.integrate(f1sp, (xsp, -1, 1))
#данная функция является плотностью распределение, т.к. f(x) > 0 и интеграл равен 1
```

```python
a = sp.symbols('a')
x = np.linspace(0, 2, 10)
f2 = np.where((x >= 0) & (x <= 2), 1*x**2, 0)
xsp = sp.symbols('x')
f2sp = sp.Piecewise((a*xsp**2, sp.And(xsp >= 0, xsp <= 2)))

plt.plot(x, f2)
plt.show()
sp.integrate(f2sp, (xsp, 0, 2))
#интеграл равен (8*a)/3
# можно предположить, что функция будет плотностью распределение при a равном 3/8)
```

## Задача 1

Известно, что случайная величина $X$ задана нормальным распределением $N(3,2)$. Найдите:

1. $p\{-3 <X<5\}$
2. $p\{X \leq 4\}$

```python
from scipy.stats import norm

gauss = norm(loc=3, scale=np.sqrt(2))
x = np.linspace(-5, 10, 100)
plt.plot(x, gauss.cdf(x))

p1 = gauss.cdf(5) - gauss.cdf(-3)
p2 = gauss.cdf(4)
display(f'для p1 {p1}')
display(f'для p2 {p2}')
```

## Задача 2

Случайная величина $Т$ — время работы радиолампы. Она имеет экспоненциальное распределение.

Найдите вероятность того, что лампа проработает не менее $800$ часов, если среднее время работы лампы — $400$. Постройте график плотности распределения и функции распределения.

```python
from scipy.stats import expon

exp_r = expon(scale = 400)
fig, ax = plt.subplots(2, 1)

p = 1-exp_r.cdf(800)
x = np.linspace(0, 1000, 1000)
ax[0].plot(x, exp_r.pdf(x))
ax[0].set_title('pdf')
ax[1].plot(x, exp_r.cdf(x))
ax[1].set_title('cdf')
plt.legend()
display(p)
```

## Задача 1

В соревнованиях по стрельбе из лука два участника. Матч длится два раунда, засчитывается последний. Каждый участник сделал по 20 выстрелов. Очки считаются по сумме выстрелов.

На основе полученных данных у первого стрелка вышло распределение Рэлея (отклонение от попадания в яблочко $10$) $Ray(0,3)$, у другого — $Ray(5,1)$. Оба лучника в отличной форме и за один раунд лучше стрелять не научатся.

У кого шанс выиграть выше и что нужно для этого сделать данному лучнику?

Необходимо:
1. Построить результаты распределения после первого раунда на одном графике.
2. Описать, кто выиграет и что для этого нужно сделать данному лучнику.
3. Проверить гипотезу численно, сгенерировав полученный результат. Для упрощения считаем, что можно получать дробные числа.

```python
from scipy.stats import rayleigh

x = np.linspace(0, 15, 1000)
arch1 = rayleigh(scale=3)
arch2 = rayleigh(scale=1)
pdf1 = arch1.pdf(x)
pdf2 = np.where(x >= 5, arch2.pdf(x-5),0)
plt.xlabel('Отклонение от центра')

plt.plot(x, pdf1, label='Лучник 1')
plt.plot(x, pdf2, label='Лучник 2')
plt.legend()
```

Выиграет лучник 1, т.к. второй не способен попадать ближе 5 от центра. Чтобы выиграть 1-му лучнику нужно просто стрелять, как он это обычно делает

```python
#0 минимальное отклонение от яблочка
# чем выше сумма тем хуже стрельба(больше отклонение)
sh1 = arch1.rvs(20)
display(sh1)
sh1.sum()
```

```python
sh2 = arch2.rvs(20)+5
display(sh2)
sh2.sum()
```

Полученные результат подтверждают гипотезу, основанную на графиках плотности вероятности
